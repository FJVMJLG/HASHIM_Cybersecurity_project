```python
# HASHIM - Cybersecurity Project
# Educational network information and security basics

import socket
import platform
from datetime import datetime

print("=" * 45)
print("       HASHIM CYBERSECURITY PROJECT")
print("=" * 45)

print("\n[1] System Information")
print("System:", platform.system())
print("Python version:", platform.python_version())
print("Computer name:", socket.gethostname())

print("\n[2] Network Information")
try:
    hostname = socket.gethostname()
    ip_address = socket.gethostbyname(hostname)
    print("Hostname:", hostname)
    print("Local network IP:", ip_address)
except Exception as error:
    print("Network information unavailable:", error)

print("\n[3] Cybersecurity Fundamentals")
topics = [
    "Use strong and unique passwords",
    "Enable two-factor authentication",
    "Keep software updated",
    "Never open suspicious links",
    "Test networks only with permission"
]

for number, topic in enumerate(topics, start=1):
    print(f"{number}. {topic}")

print("\nReport generated:", datetime.now().strftime("%Y-%m-%d %H:%M:%S"))
print("\nProject completed successfully!")
```
